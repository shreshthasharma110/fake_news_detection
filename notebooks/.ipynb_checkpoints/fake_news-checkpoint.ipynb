```python
import pandas as pd
import numpy as np
import matplotlib as plt 
import seaborn as sns
from newspaper import Article
```

```python
true = pd.read_csv("../dataset/True.csv")
fake = pd.read_csv("../dataset/Fake.csv")
```

```python
true["text"] = true["text"].replace("(Reuters)","",regex=True)
true.head()
```

```python
fake.head(5)
```

```python
true["label"]= 1
fake["label"]= 0
```

```python
fake.head()
```

```python
news = pd.concat([fake,true],axis=0)
```

```python
news.head()
```

```python

news.shape
```

```python
news = news.drop(['title', 'subject', 'date'],axis=1)
```

```python
news.head()
```

```python
news.tail()
```

```python
news['label'].value_counts()
```

## reshuffling

```python
news = news.sample(frac=1)
```

```python
news.head()
```

```python
 news.reset_index(inplace=True)
```

```python
news.head()
```

```python
news.drop(['index'],axis= 1, inplace=True)
```

```python
news.head()
```

```python
import re
```

```python
def wordopt(text):
    #Covert into lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+','',text)
    #remove html tags
    text = re.sub(r'<.*?>','',text)
    #Remove punctuation
    text = re.sub(r'[^\w\s]','',text)
    #Remove digits
    text = re.sub(r'\d','',text)
    #remove newline characters
    text = re.sub(r'\n','',text)

    return text
    
```

```python
news["text"] = news["text"].apply(wordopt)
```

```python
news['text']
```

```python
x = news['text']
y = news['label']
```

```python
x
```

```python
y
```

```python
fake.tail()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.3)
```

```python
x_train.shape
```

```python
x_test.shape
```

## feature extraction

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorization = TfidfVectorizer()
```

```python
xv_train = vectorization.fit_transform(x_train)
```

```python
xv_test = vectorization.transform(x_test)
```

```python
xv_train

```

```python
xv_test
```

# Now creating ML model

```python
from sklearn.linear_model import LogisticRegression
```

```python
LR =  LogisticRegression()
```

```python
LR.fit(xv_train , y_train)
```

```python
pred_lr = LR.predict(xv_test)
```

```python
LR.score(xv_test, y_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test ,pred_lr))
```

```python
def output_news(n):
    if n==0:
        return 'it is a fake news'
    elif n==1:
        return 'it is true news' 
```

```python
news = input("Enter the news : ")
news = wordopt(news)
news_vector = vectorization.transform([news])
prediction = LR.predict(news_vector)
print('\nPrediction: ')
print(output_news(prediction[0]))
```

# Adding url section


```python
from newspaper import Article

url = input("Enter News URL: ")

article = Article(url)

article.download()
article.parse()

news = article.text

news = wordopt(news)

news_vector = vectorization.transform([news])

prediction = LR.predict(news_vector)

print("Title:", article.title)
print("Prediction:", output_news(prediction[0]))
```
